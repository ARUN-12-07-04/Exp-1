```python
import numpy as np
arr1=np.array([10,20,30,40,50])
arr2= np.array([5,10,15,20,25])
print("array1:",arr1)
print("array2:",arr2)

print("addition of two arrays:",arr1+arr2)
print("subtraction of two arrays:",arr1-arr2)
print("multiplication of two arrays:",arr1*arr2)
print("division of two arrays:",arr1/arr2)

#step2:pandas Dataframe
import pandas as pd
student={
    'name':["john",'alice','bob','daisy'],
    'age':[25,30,22,28],
    'salary':[40000,50000,32000,45000]

}
df=pd.DataFrame(student)
#display data
print("\ndataset:")
print(df)
#descriptive statics
print("\n statistics:")
print(df.describe())
print(df.info())
```
